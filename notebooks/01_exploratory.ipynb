```python
# Make the `ark` package importable from a notebook without installing.
import sys
from pathlib import Path

sys.path.insert(0, str(Path.cwd().parent / "src"))

from ark import config, set_seed
from ark.config import FIGURES_DIR

set_seed(config.seed)
config
```

```python
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.datasets import load_wine

data = load_wine(as_frame=True)
df = data.frame
df.head()
```

```python
df.describe()
```

```python
ax = df["alcohol"].plot(kind="hist", bins=20, title="Alcohol distribution")
ax.set_xlabel("alcohol")
plt.tight_layout()
# Save publication-ready figures to results/figures/
FIGURES_DIR.mkdir(parents=True, exist_ok=True)
plt.savefig(FIGURES_DIR / "alcohol_hist.png", dpi=150)
```
